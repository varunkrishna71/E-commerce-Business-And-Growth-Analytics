## 4. Monthly Sales Analysis

### Business Question

Which month generated the highest and lowest total sales?

### Approach

The order date is used to extract the month and year.
Sales are then aggregated to compare monthly performance.

# Monthly Analysis, Year Analysis, Order Day of week mtlb delivery kab tak ho ja rhi hai

```python
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data
```

# Monthly Sales Analysis

### You need to calculate the monthly sales of the store and identify which month had the highest sales and which month had the lowest sales.

```python
sales_by_month1 = data.groupby(['Order Month','Order Year'])['Sales'].sum().reset_index()
sales_by_month1
```

```python
sales_by_month = data.groupby(['Order Month'])['Sales'].sum().reset_index()
fig = px.line(sales_by_month,x = 'Order Month', y = 'Sales',title = 'Monthly Sales Analysis')
fig.show()
```

## 5. Sales Analysis by Product Category

### Business Question

Which product category generates the highest and lowest sales?

# sales by Category

### You need to analyze sales based on product categories and determine which category has the lowest sales and which category has the highest sales.

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
fig = px.pie(sales_by_category,values='Sales',names ='Category',hole=0.3,color_discrete_sequence = px.colors.qualitative.Pastel)
fig.update_traces(textposition='inside', textinfo = 'percent+label')
fig.update_layout(title_text= 'Sales Analysis by Category', title_font = dict(size = 25))
fig.show()
```

## 8. Profitability Analysis by Category and Sub-Category

```python
fig = px.pie(profit_by_category,values = 'Profit',names = 'Category', hole = 0.3, color_discrete_sequence = px.colors.qualitative.Pastel)
fig.update_traces(textposition='inside' , textinfo = 'percent+label')
fig.update_layout(title_text='Profit Analysis by Category', title_font = dict(size=24))
```

## 9. Sales and Profit by Customer Segment

### Business Question

How do different customer segments contribute to sales and profit?

The analysis compares Consumer, Corporate, and Home Office
segments using total sales and total profit.

# Sales And Profit - Customer Segment

# Analyze the sales and Profit by Customer Segment

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
color_palette = colors.qualitative.Pastel
fig = go.Figure()
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],y=sales_profit_by_segment['Sales'],name = 'Sales', marker_color = color_palette[0]))

fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'], y = sales_profit_by_segment['Profit'],name = 'Profit',marker_color = color_palette[1]))
fig.update_layout(title = 'Sales and Profit Analysis by Customer Segment',xaxis_title = 'Customer Segment', yaxis_title = 'Amount')
fig.show()
```

## 10. Sales-to-Profit Ratio Analysis

This section examines the relationship between total sales
and total profit to understand overall profitability.

# sales to profit ratio

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum','Profit':'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales']/sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment','Sales_to_Profit_Ratio']])
```
